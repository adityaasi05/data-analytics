```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# 1. Setup Data
# Subplot 1 Data
x = [1, 2, 3, 4, 5]
y = [10, 15, 13, 17, 20]

# Subplot 2 Data
categories = ['A', 'B', 'C', 'D']
values = [23, 45, 56, 78]

# Subplot 3 Data
labels = ["Apples", "Oranges", "banana", "Datas"]
sizes = [25, 28, 35, 70]

# Subplot 4 Data (Synthetic Tips Dataset)
df = pd.DataFrame({
    'total_bill': np.random.uniform(5, 50, 244),
    'day': np.random.choice(['Thur', 'Fri', 'Sat', 'Sun'], 244),
    'sex': np.random.choice(['Male', 'Female'], 244)
})

# 2. Plotting
plt.figure(figsize=(14, 10))

# Subplot 1: Line Plot
plt.subplot(2, 2, 1)
plt.plot(x, y, marker='*', label='line plot with * marker', color='red')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.grid(True)
plt.title('Line Plot')

# Subplot 2: Bar Plot
plt.subplot(2, 2, 2)
plt.bar(categories, values, color=['red', 'blue', 'green', 'orange'], hatch="//")
plt.title('Bar Plot Example')
plt.xlabel('Categories')
plt.ylabel('Values')

# Subplot 3: Pie Chart
plt.subplot(2, 2, 3)
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140, shadow=True)
plt.title('Fruit Distribution')

# Subplot 4: Box Plot
plt.subplot(2, 2, 4)
sns.boxplot(data=df, x='day', y='total_bill', hue='sex')
plt.title('Box Plot of Total Bill by Day and Gender')

plt.tight_layout()
plt.show()
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

# Sales data for 2024 (in thousands)
sales_2024 = [50, 55, 53, 60, 65, 70, 75, 72, 68, 70, 74, 80]

# Sales data for 2025 (in thousands)
sales_2025 = [52, 58, 56, 62, 70, 75, 78, 76, 73, 77, 79, 85]

# Create the line plot
plt.plot(months, sales_2024, label='Sales 2024', color='blue', marker='o')
plt.plot(months, sales_2025, label='Sales 2025', color='green', marker='s')

# Add labels and title
plt.xlabel('Month')
plt.ylabel('Sales (in thousands)')
plt.title('Monthly Sales Trends: 2024 vs 2025')

# Show legend
plt.legend()

# Display the plot
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

sales_2024 = [50, 55, 53, 60, 65, 70, 75, 72, 68, 70, 74, 80]
sales_2025 = [52, 58, 56, 62, 70, 75, 78, 76, 73, 77, 79, 85]

# Build a DataFrame of it.
data = pd.DataFrame({
    'Month': months * 2,
    'Sales': sales_2024 + sales_2025,
    'Year': ['2024'] * 12 + ['2025'] * 12
})
```

```python
data.head()
```

```python
# Set the style and create line plot
sns.set(style="whitegrid")

plt.figure(figsize=(10, 6))
sns.lineplot(data=data, x='Month', y='Sales', hue='Year', marker='o')

#Title 
#xlabel
#ylabel

plt.title('Monthly Sales Trends: 2024 vs 2025')
plt.xlabel('Month')
plt.ylabel('Sales (in thousands)')

# Show the plot
plt.tight_layout()
plt.show()
```

```python
import plotly.express as px
import pandas as pd

# Sample Titanic data
data = {
    "Sex": ["male", "male", "female", "female"],
    "Status": ["Survived", "Dead", "Survived", "Dead"],
    "Count": [100, 500, 200, 80]
}

df = pd.DataFrame(data)

# bar chart
fig = px.bar(
    df,
    x="Sex",
    y="Count",
    color="Status",          # color by survived/dead
    barmode="group",         # group bars side by side
    title="Titanic Survival by Gender"
)

# Show chart
fig.show()
```
